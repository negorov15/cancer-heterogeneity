```python
import os
import pandas as pd
```

```python
cm_dir = "input/count_matrices"
```

```python
LINES = pd.read_csv("input/cell_lines_modelids.tsv", sep="\t")
CELL_LINES = LINES["Cell line"].to_list()
```

```python
for cl in CELL_LINES:
    files = os.listdir(cm_dir)
    if len(files) == 0:
        print(f"No files found in directory for {cl}.")
    cm = pd.read_csv(cm_dir + "/" + cl + ".count_mtx.tsv.txt", sep="\t", index_col=0)
    # filter out genes with zero counts across all cells
    cm = cm.loc[cm.sum(axis=1) > 0]
    print(f"Count matrix for {cl} has {cm.shape[0]} genes and {cm.shape[1]} cells.")
```
