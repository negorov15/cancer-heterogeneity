```python
import os
import sys
import pandas as pd
```

```python
wd = "/Users/nikoegorov/cancer-heterogeneity/"
# Input files and directories
MUTATIONS = f"{wd}/input/OmicsSomaticMutations.csv"
MUTATIONS = pd.read_csv(MUTATIONS, sep=",")
LINES_TSV = f"{wd}/input/cell_lines_modelids.tsv"
LINES = pd.read_csv(LINES_TSV,sep="\t")
LINES.head()
EXPR_DIR  = f"{wd}/input/ExpressionData"             # Contains total UMIs per gene per cell line
```

```python
display(MUTATIONS.head())
MUTATIONS.info()
```

```python
# keep only default entries for models
mutations = MUTATIONS[(MUTATIONS["IsDefaultEntryForModel"] == "Yes") & 
                      (MUTATIONS["VariantType"] == "SNV")]
display(mutations.info())
```

```python
CELL_LINE = LINES["Cell line"][0]   # Take A_253 as example
CL_EXPRESSION_DIR = f"{EXPR_DIR}/{CELL_LINE}/{CELL_LINE}_genes_sorted_by_expression.txt"
CL_EXPRESSION = pd.read_csv(CL_EXPRESSION_DIR, sep="\t", header=None, names=["HugoSymbol", "Expression"])
CL_EXPRESSION.head()
```

```python
CL_MODEL_ID = LINES["Model_ID"][0]
CL_MUTATIONS = mutations[mutations["ModelID"]==CL_MODEL_ID]
CL_MUTATIONS.info()
```

```python
expressed_genes= CL_EXPRESSION[CL_EXPRESSION["Expression"] > 0]
expressed_hugo_symbols = expressed_genes["HugoSymbol"].tolist()
variants_expressed = CL_MUTATIONS[CL_MUTATIONS["HugoSymbol"].isin(expressed_hugo_symbols)]
variants_expressed.head()
```

```python
columns = list(variants_expressed.columns)
columns_to_keep = ["ModelID", "Chrom", "Pos", "Ref", "Alt", "HugoSymbol"]
columns_to_drop = [col for col in columns if col not in columns_to_keep]
variants_expressed.drop(columns_to_drop, axis=1)
```
